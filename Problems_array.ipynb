```python
# using hash maps 

def word_to_number(word):
    # Hash map for basic numbers
    word_map = {
        "zero": 0, "one": 1, "two": 2, "three": 3, "four": 4, "five": 5,
        "six": 6, "seven": 7, "eight": 8, "nine": 9, "ten": 10,
        "eleven": 11, "twelve": 12, "thirteen": 13, "fourteen": 14,
        "fifteen": 15, "sixteen": 16, "seventeen": 17, "eighteen": 18,
        "nineteen": 19, "twenty": 20, "thirty": 30, "forty": 40,
        "fifty": 50, "sixty": 60, "seventy": 70, "eighty": 80, "ninety": 90
    }
    # Hash map for multipliers
    multiplier_map = {"hundred": 100, "thousand": 1000, "million": 10**6}

    words = word.replace("-", " ").split()  # Split the words by spaces
    total = 0
    current = 0

    for w in words:
        if w in word_map:  # Basic numbers
            current += word_map[w]
        elif w in multiplier_map:  # Multipliers
            current *= multiplier_map[w]
            if w != "hundred":  # Add only for 'thousand' and 'million'
                total += current
                current = 0
        elif w == "and":  # Skip 'and'
            continue

    total += current  # Add any remaining value
    return total

# Example usage
print(word_to_number("two hundred and fifty-five"))  # Output: 255
print(word_to_number("three thousand and twenty"))   # Output: 3020
print(word_to_number("one million two hundred"))     # Output: 1000200
```

```python
# # Intermediate Problems (Performing Operations)
# Sorting and Searching:

# Sort an array [4, 2, 7, 1, 9] in ascending order.
# Write a program to check if a given element (e.g., 7) 
# is present in the array using binary search. Assume the array is already sorted.

arr = [4, 2, 7, 1, 9]
arr.sort()
print(f"Sorted array is : {arr}")
def binary_search(arr,target):
    low = 0
    high = len(arr)-1
    while low <= high:
        mid = (low + high)//2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1
# Test the binary search function
target = int(input("Enter the number to find :"))
result = binary_search(arr,target)
if result != -1:
    print(f" Element {target} found at index : {result}")

else:
    print(f"Element {target} is not found")
```

```python
# Write a function to merge two sorted arrays into one sorted array.
# Example:
# python
# Copy code
# arr1 = [1, 3, 5]
# arr2 = [2, 4, 6]
# Result: [1, 2, 3, 4, 5, 6]

def merge_sort(arr1,arr2):
    arr1.extend(arr2)
    arr1.sort()
    return arr1

# Its gives the output in sorted order but we can also use merge sort algorithm to merge the two sorted arrays
# Test the function
arr1 = [1, 3, 5]
arr2 = [2, 4, 6]
print(merge_sort(arr1,arr2))
```

```python
# Finding the Maximum and Minimum:

# Write a program to find the maximum and minimum values in an array without using built-in functions.

def find_max_min(arr):
    max_val = arr[0]
    min_val = arr[0]
    for num in arr:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num

    return max_val, min_val
# Test the function
arr = [10, 20, 30, 40, 50]
result = find_max_min(arr)
print(result)
```

```python
# find the index of the maximum value in the array
def find_index_max(arr):
    n = len(arr)
    for i in range(n-1):
        max_value = i
        for j in range(i+1,n):
            if arr[j] > arr[max_value]:
                max_value = j
    return max_value

arr = [10, 20, 30, 40, 50]
print("Index of maximum element is", find_index_max(arr))
```

```python
# Write a function to find duplicate elements in an array. Example:
# python
# Copy code
# arr = [1, 2, 3, 2, 4, 5, 1]
# Result: [1, 2]
# Method 1: using modules
from collections import Counter
from typing import List
def find_duplicates_fun(arr: List[int]) -> List[int]:
    # Create a Counter object from the array
    counter = Counter(arr)
    # Extract elements with count > 1
    duplicates = [item for item, count in counter.items() if count > 1]
    return duplicates
# method 2: using nested loops
def find_duplicates(arr):
    duplicates =[]
    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            if arr[i] == arr[j]:
                duplicates.append(arr[i])
                
    return duplicates
# method 3: using count method
def find_duplicates_count(arr):
    duplicates =[]
    for i in arr:
        if (arr.count(i) > 1) and (i not in duplicates):
            duplicates.append(i)
    return duplicates

# Test the function
arr = [1, 2, 3, 2, 4, 5, 1]
print(find_duplicates_fun(arr))
print(find_duplicates(arr))
print(find_duplicates_count(arr))
```

```python
# Advanced Problems (Algorithmic Challenges)( This is leetcode problem)

# Subarray with a Given Sum:

# Find a subarray with a given sum in an array. Example:
# python
# Copy code
# arr = [1, 4, 20, 3, 10, 5]
# Target sum = 33
# Result: [20, 3, 10]

def three_no_Sum(nums = list[int], target = int):
    n = len(nums)
    for i in range (n-2):
        for j in range (i+1, n-1):
            for k in range (j+1, n):
                if nums[i] + nums[j] + nums[k] == target:
                    return [nums[i], nums[j], nums[k]]
                
    return []

def subarraySum(arr, target):
    n = len(arr)
    for i in range(n):
        current_sum = arr[i]
        j = i + 1
        while j <= n:
            if current_sum == target:
                return arr[i:j]
            if current_sum > target or j == n:
                break
            current_sum += arr[j]
            j += 1
    return []

# Test the function
arr = [1, 4, 20, 3, 10, 5]
target = 33
print(subarraySum(arr, target))
```

```python
# Rearrange Positive and Negative Numbers:

# Rearrange an array such that all negative numbers come before positive numbers while maintaining their relative order. Example:
# python
# Copy code
# Input: [1, -2, 3, -4, -5, 6]
# Output: [-2, -4, -5, 1, 3, 6]
def rearrange(arr):

    n = len(arr)
    j = 0
    for i in range(n):
        if arr[i] < 0:
            arr[i], arr[j] = arr[j], arr[i]
            j += 1
    return arr


# using lambda function
arr = [1, -2, 3, 4, -5, 6]
arr.sort(key = lambda x: 0 if x < 0 else 1)
print(arr)
```

```python
# Finding the Majority Element:

# A majority element is an element that occurs more than n/2 times in an array of size n. Write a function to find it. Example:
# python
# Copy code
# arr = [3, 3, 4, 2, 3, 3, 5]
# Output: 3
def majority_element(arr):
    n = len(arr)
    for i in range(n):
        count = 0
        for j in range(n):
            if arr[i] == arr[j]:
                count +=1
            if count > n//2:
                return arr[i]
    return "No majority element found"       

# Test the function
arr = [3, 3, 4, 3,3,2,5]
print(majority_element(arr))
```

```python
# Product of Array Except Self:

# Write a program to find the product of all elements in an array except the current element without using division. Example:
# python
# Copy code
# arr = [1, 2, 3, 4]
# Result: [24, 12, 8, 6]

def product_except_self(arr):
    n = len(arr)
    result = [1]*n
    left = 1
    for i in range(n):
        result[i] *= left
        left *= arr[i]

    right = 1   
    for i in range(n-1,-1,-1):
        result[i] *= right
        right *= arr[i]
    return result
# Test the function
arr = [1, 2, 3, 4]
print(product_except_self(arr))
```

```python
def partition_array(arr, pivot):
    less_than = [x for x in arr if x < pivot]
    equal_to = [x for x in arr if x == pivot]
    greater_than = [x for x in arr if x > pivot]
    return less_than + equal_to + greater_than

# Example usage
arr = [9, 12, 3, 5, 14, 10, 10]
pivot = 10
print(partition_array(arr, pivot))  # Output: [9, 3, 5, 10, 10, 12, 14]
```

```python
def find_median_sorted_arrays(nums1, nums2):
    nums = sorted(nums1 + nums2)
    n = len(nums)

    if n % 2 == 1:  # Odd length
        return nums[n // 2]
    else:  # Even length
        return (nums[(n - 1) // 2] + nums[n // 2]) / 2

# Example usage
nums1 = [1, 3]
nums2 = [2]
print(find_median_sorted_arrays(nums1, nums2))  # Output: 2
```
